```python
# HANDWRITTEN DIGIT AND ALPHABET RECOGNITION USING CNN
# Google Colab - Single Cell Complete Project

!pip -q install tensorflow-datasets importlib_resources

import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
from google.colab import files
import io

# 1. LOAD DIGIT DATASET
print("Loading MNIST digits...")
(xd_train, yd_train), (xd_test, yd_test) = (
    tf.keras.datasets.mnist.load_data()
)

# 2. LOAD ALPHABET DATASET
print("Loading EMNIST letters...")
train_ds, test_ds = tfds.load(
    "emnist/letters",
    split=["train", "test"],
    as_supervised=True
)

def load_letters(dataset):
    images, labels = [], []
    for image, label in tfds.as_numpy(dataset):
        images.append(image.squeeze())
        labels.append(int(label))
    images = np.array(images)
    images = np.transpose(images, (0, 2, 1))
    return images, np.array(labels)

xl_train, yl_train = load_letters(train_ds)
xl_test, yl_test = load_letters(test_ds)

# Digit classes: 0-9; letter classes: 10-35
yl_train = yl_train + 9
yl_test = yl_test + 9

# 3. COMBINE BOTH DATASETS
x_train = np.concatenate([xd_train, xl_train])
y_train = np.concatenate([yd_train, yl_train])
x_test = np.concatenate([xd_test, xl_test])
y_test = np.concatenate([yd_test, yl_test])

# 4. PREPROCESS IMAGES
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train[..., np.newaxis]
x_test = x_test[..., np.newaxis]

# Shuffle training data
rng = np.random.default_rng(42)
idx = rng.permutation(len(x_train))
x_train, y_train = x_train[idx], y_train[idx]

# 5. BUILD CNN MODEL
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28, 28, 1)),
    tf.keras.layers.Conv2D(32, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(64, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(36, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 6. TRAIN MODEL
print("Training CNN model...")
model.fit(
    x_train, y_train,
    epochs=2,
    batch_size=128,
    validation_split=0.1
)

# 7. EVALUATE MODEL
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"\nTest accuracy: {accuracy * 100:.2f}%")

# Save trained model
model.save("/content/handwriting_model.keras")
print("Model saved!")

# 8. UPLOAD HANDWRITTEN IMAGE AND PREDICT
print("\nUpload an image containing ONE handwritten digit or capital letter.")
uploaded = files.upload()

for filename, content in uploaded.items():
    image = Image.open(io.BytesIO(content)).convert("L")

    # Convert dark writing on white paper to white on black
    image = ImageOps.invert(image)

    # Crop around the handwritten character
    image = image.crop(image.getbbox())

    if image.width == 0 or image.height == 0:
        print("Blank image. Please upload another image.")
        continue

    # Preserve proportions and fit into 20x20 area
    image.thumbnail((20, 20))

    canvas = Image.new("L", (28, 28), 0)
    x_offset = (28 - image.width) // 2
    y_offset = (28 - image.height) // 2
    canvas.paste(image, (x_offset, y_offset))

    arr = np.array(canvas, dtype=np.float32) / 255.0
    arr = arr.reshape(1, 28, 28, 1)

    # Predict character
    predictions = model.predict(arr, verbose=0)[0]
    predicted_class = int(np.argmax(predictions))

    if predicted_class < 10:
        character = str(predicted_class)
    else:
        character = chr(ord("A") + predicted_class - 10)

    confidence = predictions[predicted_class] * 100

    print("\n========== RESULT ==========")
    print("Image:", filename)
    print("Predicted Character:", character)
    print(f"Confidence: {confidence:.2f}%")

    plt.figure(figsize=(4, 4))
    plt.imshow(canvas, cmap="gray")
    plt.title(f"Prediction: {character} ({confidence:.2f}%)")
    plt.axis("off")
    plt.show()

print("\nProject completed successfully!")
```
